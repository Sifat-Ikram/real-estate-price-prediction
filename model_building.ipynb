```python
def is_float(x):
    try:
        float(x)
    except:
        return False
    return True
df1[~df1['total_sqft'].apply(is_float)]
```

```python
def convert_to_num(x):
    token = x.split('-')
    if len(token) == 2:
        return (float(token[0]) - float(token[1]))/2
    try:
        return float(x)
    except:
        return None
```

```python
convert_to_num('21')
```

```python
convert_to_num('1200 - 1470')
```

```python
convert_to_num('34Sq. Meter')
```

```python
df1['total_sqft'] = df1['total_sqft'].apply(convert_to_num)
```
